```python
import json 
import matplotlib.pyplot as plt 

json_path = "Trajectories/track_monteblanco.json"
```

```python
# draw the trajectory from the json file
with open(json_path, 'r') as f:
    data = json.load(f)


# data.keys() = dict_keys(['X', 'Y', 'X_i', 'Y_i', 'X_o', 'Y_o'])

# plot the trajectory
plt.figure(figsize=(10, 10))
# plot the inner and outer boundaries
plt.plot(data['X_i'], data['Y_i'], label='Inner Boundary', color='red')
plt.plot(data['X_o'], data['Y_o'], label='Outer Boundary', color='green')
plt.plot(data['X'], data['Y'], label='Trajectory', color='blue')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Trajectory')
plt.legend()
plt.show()
```
